```python
train = pd.read_csv('../data/diamonds_train.csv')
train.head()
```

```python
test = pd.read_csv('../data/diamonds_test.csv')
test.head()
```

<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: right;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش/مهندسی ویژگی که در فصل‌های گذشته آموختید، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.
</font>
</p>

```python
train.isna().sum()
```

```python
train.cut.value_counts()
train.color.value_counts()
```

```python
train.replace({'cut': {'Fair': 0, 'Good': 1, 'Very Good': 2, 'Premium': 3,  'Ideal': 4},
               'color': {'J': 0, 'I': 1, 'H': 2, 'G': 3, 'F': 4, 'E': 5, 'D': 6},
               'clarity': {'I1': 0, 'SI2': 1, 'SI1': 2, 'VS2': 3, 'VS1': 4, 'VVS2': 5, 'VVS1': 6, 'IF': 7}}, inplace=True)
train
```

```python
train.describe()
```

```python
Q1 = train['x'].quantile(0.25)
Q3 = train['x'].quantile(0.75)
IQR = Q3 - Q1

train = train[~((train['x'] < (Q1 - 1.5 * IQR)) | (train['x'] > (Q3 + 1.5 * IQR)))]
```

```python
Q1 = train['y'].quantile(0.25)
Q3 = train['y'].quantile(0.75)
IQR = Q3 - Q1

train = train[~((train['y'] < (Q1 - 1.5 * IQR)) | (train['y'] > (Q3 + 1.5 * IQR)))]
```

```python
Q1 = train['z'].quantile(0.25)
Q3 = train['z'].quantile(0.75)
IQR = Q3 - Q1

train = train[~((train['z'] < (Q1 - 1.5 * IQR)) | (train['z'] > (Q3 + 1.5 * IQR)))]
```

```python
Q1 = train['depth'].quantile(0.25)
Q3 = train['depth'].quantile(0.75)
IQR = Q3 - Q1

train = train[~((train['depth'] < (Q1 - 1.5 * IQR)) | (train['depth'] > (Q3 + 1.5 * IQR)))]
```

```python
Q1 = train['table'].quantile(0.25)
Q3 = train['table'].quantile(0.75)
IQR = Q3 - Q1

train = train[~((train['table'] < (Q1 - 1.5 * IQR)) | (train['table'] > (Q3 + 1.5 * IQR)))]
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
train[['x', 'y', 'z', 'depth', 'table']] =  scaler.fit_transform(train[['x', 'y', 'z', 'depth', 'table']])
train
```

    الگوریتم <code>LinearRegression</code> نیز در کتابخانه‌ی <code>scikit-learn</code> به‌خوبی پیاده‌سازی شده و می‌توانیم از آن جهت اجرای الگوریتم <i>رگرسیون خطی</i> بر روی داده‌های خود کمک بگیریم.

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
آموزش مدل
</font>
</h3>

<p dir=rtl style="direction: rtl;text-align: right;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    در سلول پایین با استفاده از کتابخانه <code>scikit-learn</code> مدلی آموزش دهید که بتواند متغیر هدف این مسئله را پیش‌بینی کند.
    <br>
    <b>حتما باید از مدل رگرسیون خطی یعنی <code>LinearRegression</code> استفاده کنید.</b>
</font>
</p>

```python
X = train.drop('price', axis=1)
y = train['price']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test,  y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
y_predict = model.predict(X_test)
```

<h3 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
معیار ارزیابی

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    معیاری که برای ارزیابی عملکرد مدل انتخاب کرده‌ایم، <code>r2_score</code> نام دارد.
    <br>
    این معیار، سنجه ارزیابی کیفیت مدل شماست. به عبارت بهتر در سامانه داوری هم از همین معیار برای نمره‌دهی استفاده شده است.
    <br>
    پیشنهاد می‌شود با توجه به این معیار، عملکرد مدل خود را بر روی مجموعه‌ی آموزش یا اعتبارسنجی ارزیابی کنید.

```python
from sklearn.metrics import r2_score

accurecy = r2_score(y_test, y_predict)
accurecy
```
